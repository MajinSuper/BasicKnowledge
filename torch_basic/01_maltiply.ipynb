```python
import torch
x_2d = torch.rand([64,128])
y_2d = torch.rand([64,128])

x_3d = torch.rand([64,100,128])
y_3d = torch.rand([64,100,128])

vector = torch.rand([128,])
```

# 1. torch.mm

- 只能二维矩阵相乘




```python
res = torch.mm(x_2d,y_2d.transpose(-1,-2))
print(res.shape)
```

```python
res = torch.mm(x_3d,y_3d.transpose(-1,-2))
res.shape
```

# 2. torch.bmm
- 只能三维矩阵相乘

```python
res = torch.bmm(x_2d,y_2d.transpose(-1,-2))
print(res.shape)
```

```python
res = torch.bmm(x_3d,y_3d.transpose(-1,-2))
res.shape
```

# 3. torch.matmal
- 多维矩阵相乘都可以
- \>2维，只用最后两维度
- 等价于 @

```python
res = torch.matmul(x_2d,y_2d.transpose(-1,-2))
print(res.shape)
```

```python
res = torch.matmul(x_3d,y_3d.transpose(-1,-2))
print(res.shape)
```

```python
res = x_2d @ y_2d.transpose(-1,-2)
print(res.shape)
```

```python
res = x_3d @ y_3d.transpose(-1,-2)
print(res.shape)
```

# 4.torch.mul
- 元素维度 对应乘
- 等价于 * 

```python
res1 = torch.mul(x_2d,y_2d)
res2 = x_2d * y_2d

print(res1.shape)
print(res2.shape)
torch.equal(res1,res2)
```

```python
res1 = torch.mul(x_3d,y_3d)
res2 =  x_3d * y_3d

print(res1.shape)
print(res2.shape)
torch.equal(res1,res2)
```

# 5.torch.mv
- 矩阵乘以向量
- 等价于@

```python
res1 = x_2d @ vector
res2 = torch.mv(x_2d,vector)
print(res1.shape)
print(res2.shape)
print(torch.equal(res1,res2))
```

# 6. torch.dot
- 向量点乘
- 等价于@

```python
res1 = vector @ vector
res1
```

```python
res2 = torch.dot(vector,vector)
res2
```
